## Working with Missing Data in Pandas

```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

### Filling missing values using fillna(), replace() and interpolate()

```python
data = {'name' : ['steve', 'john', 'mike', 'susan', 'jane', 'lisa', 'messi'],
        'age' : [20, 22, 21, 23, 22, 20, 21], 
        'gender' : ['male', 'male', 'male', 'female', 'female', 'female', 'male'],
        'rank' : [2, 1, 4, 5, 3, 6, 7]}
rank_df = DataFrame(data)
rank_df.iloc[2:5,1] = np.nan
rank_df.iloc[3:6,3] = np.nan
rank_df.iloc[3,:] = np.nan
rank_df
```

```python
rank_df.isnull()
```

```python
rank_df.notnull()
```

```python
bool_series = pd.isnull(rank_df["age"])
rank_df[bool_series]
```

```python
rank_df.fillna(0)
```

```python
rank_df.fillna(method='pad')
```

```python
rank_df.fillna(method='bfill')
```

```python
rank_df.interpolate(method='linear')
```

```python
rank_df.dropna()
```

```python
rank_df.dropna(how='all')
```

```python
rank_df.dropna(axis=1)
```

```python
rank_df.dropna(axis=0)
```
