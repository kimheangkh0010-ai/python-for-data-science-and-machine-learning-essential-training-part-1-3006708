```python
import pandas as pd
import numpy as np

from pandas import DataFrame
```

```python
number_df = pd.DataFrame(np.arange(0, 90, 3).reshape(10, 3), index=['row1', 'row2', 'row3', 'row4', 'row5', 'row6', 'row7', 'row8', 'row9', 'row10'], columns=['col1', 'col2', 'col3'])
```

#### Comparison operators (> < = <= => == !=) and Masking.

```python
number_df
```

```python
number_df.iloc[0,1]
```

```python
number_df.iloc[0,1]=20
```

```python
number_df
```

```python
number_df.iloc[[1,2,4], [1,2]]
```

```python
mask=number_df > 30
mask
```

```python
number_df[mask]
```

```python
number_df[number_df > 30] = 0
number_df
```
