```python
import matplotlib.pyplot as plt
import seaborn as sns

from pandas import DataFrame
```

```python
data = {'names': ['Bob', 'Jessica', 'Mary', 'John', 'Mel', 'Steven', 'Mob'],
        'ages': [23, 21, 22, 30, 29, 25, 21],
        'gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Male'],
        'rank': [2, 1, 3, 4, 7, 6, 5]}
df = DataFrame(data)
df
```

### Matplotlib's Bar Chart

```python
plt.bar(df['names'], df['ages'])
plt.xlabel('Names')
plt.ylabel('Ages')
plt.title('Comparison of Ages by Names')
plt.show()
```

```python
plot = sns.barplot(data=df, x='names', y='ages')
plot.set_title('Comparison of Ages by Names')
plt.show()
```

### Line Plot Matplotlib

```python
plt.plot(df['names'], df['ages'])
plt.xlabel('Names')
plt.ylabel('Ages')
plt.title('Comparison of Ages by Names')
plt.show()
```

### Line Plot Seaborn

```python
plot = sns.lineplot(data=df, x='names', y='ages')
plot.set_title('Comparison of Ages by Names')
plt.show()
```

### Pie Chart Matplotlib

```python
plt.pie(df['ages'], labels=df['names'])
plt.title('Comparison of Ages by Names')
plt.show()
```

### Pie Chart Seaborn

```python
color = sns.color_palette('pastel')[0:5]
plt.pie(df['ages'], labels=df['names'], colors=color)
plt.title('Comparison of Ages by Names')
plt.show()
```
