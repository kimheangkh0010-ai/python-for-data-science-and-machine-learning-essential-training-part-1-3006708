### Defining plot color

```python
x = range (1, 10)
y = [1, 2, 3, 4, 0.5, 4, 3, 2, 1]

plt.bar(x, y)
```

```python
wide = [0.5, 0.5, 0.5, 0.9, 0.9, 0.9, 0.5, 0.5, 0.5]
colors = ['salmon']

plt.bar(x, y, width=wide, color=colors, align='center')
```
